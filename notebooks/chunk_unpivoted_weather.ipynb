```python
# chunk_unpivoted.py
# Splits every unpivoted weather condition CSV into smaller chunk files

import csv
from pathlib import Path

BASE = Path(r"C:\ProgramData\MySQL\MySQL Server 8.0\Uploads\energy_project_data")
IN_DIR = BASE / "unpivot_all"
OUT_DIR = BASE / "chunks"
OUT_DIR.mkdir(exist_ok=True)

CHUNK_SIZE = 500000  # can be adjusted if needed

def chunk_file(path):
    with path.open("r", encoding="utf-8", newline="") as f:
        reader = csv.reader(f)
        header = next(reader)

        part = 1
        rows = []

        for row in reader:
            rows.append(row)
            if len(rows) >= CHUNK_SIZE:
                out = OUT_DIR / f"{path.stem}.part{part}.csv"
                with out.open("w", encoding="utf-8", newline="") as o:
                    w = csv.writer(o)
                    w.writerow(header)
                    w.writerows(rows)
                print(f"✓ Wrote {out}")
                part += 1
                rows = []

        if rows:
            out = OUT_DIR / f"{path.stem}.part{part}.csv"
            with out.open("w", encoding="utf-8", newline="") as o:
                w = csv.writer(o)
                w.writerow(header)
                w.writerows(rows)
            print(f"✓ Wrote {out}")

def main():
    for csv_file in IN_DIR.glob("unpivot_*.csv"):
        print(f"Chunking {csv_file.name}...")
        chunk_file(csv_file)

if __name__ == "__main__":
    main()
```
